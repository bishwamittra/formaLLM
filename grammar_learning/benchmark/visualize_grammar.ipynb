```python
def get_data(grammar_name):
    training_size = 5000
    if os.path.isfile(f"../data/{grammar_name}/sequences_w_edit_distance_{grammar_name}_{training_size * 2}_5.pkl"):
        with open(f"../data/{grammar_name}/sequences_w_edit_distance_{grammar_name}_10000_5.pkl", 'rb') as f:
            data = pickle.load(f)
            assert "train_sequences" in data
            training_size_original = len(data["train_sequences"])
            for key in data:
                print(key, len(data[key]), len(set(data[key])))
                for i, sequence in enumerate(data[key]):
                    assert sequence == data[key][i]

    return data


data = get_data(grammar_name)
```

```python
def get_meta_data(grammar_name):
    training_size = 5000
    if os.path.isfile(f"../data/{grammar_name}/meta_data_{grammar_name}_{training_size * 2}_5.pkl"):
        with open(f"../data/{grammar_name}/meta_data_{grammar_name}_{training_size * 2}_5.pkl", 'rb') as f:
            meta_data = pickle.load(f)
    return meta_data


meta_data = get_meta_data(grammar_name)
meta_data.keys()
```
